## Train the model

```python
from ultralytics import YOLO

# Initialize the model with its YAML configuration to ensure the architecture is correctly defined.
# Then load the pre-trained weights into this architecture.
model = YOLO('yolov8n.yaml')
model.load('yolov8n.pt')

model.train(
    data="data.yaml",
    epochs=50,
    imgsz=640,
    batch=16,
    project="runs",
    name="face_mask_v2"
)
```

## Evalutate model

```python
metrics = model.val()

print(metrics)
```
