## Train the model

```python
from ultralytics import YOLO

# Build from YAML and transfer pretrained weights
model = YOLO("yolo26n.yaml").load("yolo26n.pt")  # or simply: model = YOLO("yolo26n.pt")

model.train(
    data="data.yaml",
    epochs=50,
    imgsz=640,
    batch=16,
    project="runs",
    name="face_mask_v2",
)
```

## Evalutate model

```python
metrics = model.val()

print(metrics)
```
